```python
df_4302 = pd.read_csv("../datasets/schaefcomb_wang2023SimpleCorrMatrix_ds004302.tsv", sep = "\t")
```

```python
# Création de la colonne diagnosis
df_4302["diagnosis"] = df_4302["group"].map({
    "HC": "CONTROL",
    "AVH-": "SCHZ",
    "AVH+": "SCHZ",
})

df_4302["diagnosis"] = df_4302["diagnosis"].fillna("UNKNOWN")

# Renommer les valeurs de gender
df_4302["gender"] = df_4302["gender"].replace({
    "male": "M",
    "female": "F",
})

# Replacer la colonne diagnosis juste après "group"
cols = list(df_4302.columns)
group_idx = cols.index("group")
# retirer puis réinsérer à la bonne position
cols.insert(group_idx + 1, cols.pop(cols.index("diagnosis")))
df_4302 = df_4302[cols]
```

```python
df_4302
```

```python
# Ajouter la colonne site_id avant la concaténation
df_0030.loc[:, "site_id"] = "ds000030"
df_4302.loc[:, "site_id"] = "ds004302"

# Retirer group dans df_4302
df_4302 = df_4302.drop(columns=["group"])

# Retirer les sujets BIPOLAR et ADHD dans df_0030
df_0030 = df_0030[~df_0030["diagnosis"].isin(["BIPOLAR", "ADHD"])]

# Concaténer verticalement
df_all = pd.concat([df_0030, df_4302], ignore_index=True)

# Replacer site_id juste après diagnosis
cols = list(df_all.columns)
diag_idx = cols.index("diagnosis")
cols.insert(diag_idx + 1, cols.pop(cols.index("site_id")))
df_all = df_all[cols]
```

```python
### version sans PCA

import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, roc_auc_score
from tqdm import tqdm

print("Préparation des données...")

# Toutes les colonnes corr_*
all_corr_cols = [c for c in df_all.columns if c.startswith("corr_")]
X_full = df_all[all_corr_cols]

# Garder seulement les colonnes avec au moins une valeur non-NaN
valid_mask = X_full.notna().any(axis=0)
corr_cols = list(X_full.columns[valid_mask])
X = X_full[corr_cols].values

# y = 1 pour SCHZ, 0 pour CONTROL
y = (df_all["diagnosis"] == "SCHZ").astype(int).values

print(f"Nombre de sujets : {len(y)}")
print(f"Nombre total de features corr_* : {len(all_corr_cols)}")
print(f"Nombre de features après retrait des colonnes 100% NaN : {len(corr_cols)}")
print(f"Répartition SCHZ = {y.sum()}, CONTROL = {len(y) - y.sum()}")

# 10 splits stratifiés, seed 42
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

clf = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler(with_mean=True, with_std=True)),
    ("logreg", LogisticRegression(
        penalty="l2",
        solver="saga",
        max_iter=5000,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1,
    )),
])

accs = []
aucs = []

print("\nDémarrage du cross-validation 10 splits...\n")

for split_id, (train_idx, test_idx) in enumerate(tqdm(cv.split(X, y), total=10), start=1):

    print(f"\n=== Split {split_id} / 10 ===")

    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    print("  - Entraînement du modèle (imputation + scaling + logreg)...")
    clf.fit(X_train, y_train)

    print("  - Prédiction...")
    y_pred = clf.predict(X_test)
    y_proba = clf.predict_proba(X_test)[:, 1]

    acc = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba)

    accs.append(acc)
    aucs.append(auc)

    print(f"  → Accuracy split {split_id} : {acc:.3f}")
    print(f"  → AUC split {split_id}      : {auc:.3f}")

print("\n===== Résultats finaux (moyenne sur 10 splits) =====")
print(f"Accuracy : mean = {np.mean(accs):.3f}, std = {np.std(accs):.3f}")
print(f"AUC      : mean = {np.mean(aucs):.3f}, std = {np.std(aucs):.3f}")
```

```python
### verison avec PCA

import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, roc_auc_score
from tqdm import tqdm

print("Préparation des données...")

# Toutes les colonnes corr_*
all_corr_cols = [c for c in df_all.columns if c.startswith("corr_")]
X_full = df_all[all_corr_cols]

# Garder seulement les colonnes avec au moins une valeur non-NaN
valid_mask = X_full.notna().any(axis=0)
corr_cols = list(X_full.columns[valid_mask])
X = X_full[corr_cols].values

# y = 1 pour SCHZ, 0 pour CONTROL
y = (df_all["diagnosis"] == "SCHZ").astype(int).values

print(f"Nombre de sujets : {len(y)}")
print(f"Nombre total de features corr_* : {len(all_corr_cols)}")
print(f"Nombre de features après retrait des colonnes 100% NaN : {len(corr_cols)}")
print(f"Répartition SCHZ = {y.sum()}, CONTROL = {len(y) - y.sum()}")

# 10 splits stratifiés, seed 42
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

clf_pca = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler(with_mean=True, with_std=True)),
    ("pca", PCA(
        n_components=100,
        svd_solver="randomized",
        random_state=42,
    )),
    ("logreg", LogisticRegression(
        penalty="l2",
        solver="saga",
        max_iter=5000,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1,
    )),
])

accs = []
aucs = []

print("\nDémarrage du cross-validation 10 splits (avec PCA 100 composantes)...\n")

for split_id, (train_idx, test_idx) in enumerate(tqdm(cv.split(X, y), total=10), start=1):

    print(f"\n=== Split {split_id} / 10 ===")

    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    print("  - Entraînement du modèle (imputation + scaling + PCA + logreg)...")
    clf_pca.fit(X_train, y_train)

    print("  - Prédiction...")
    y_pred = clf_pca.predict(X_test)
    y_proba = clf_pca.predict_proba(X_test)[:, 1]

    acc = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba)

    accs.append(acc)
    aucs.append(auc)

    print(f"  → Accuracy split {split_id} : {acc:.3f}")
    print(f"  → AUC split {split_id}      : {auc:.3f}")

print("\n===== Résultats finaux (moyenne sur 10 splits, PCA 100) =====")
print(f"Accuracy : mean = {np.mean(accs):.3f}, std = {np.std(accs):.3f}")
print(f"AUC      : mean = {np.mean(aucs):.3f}, std = {np.std(aucs):.3f}")
```

```python
import pandas as pd
import numpy as np
import os
import warnings
from tqdm import tqdm
from pynm.pynm import PyNM

# --- RPY2 ---
import rpy2.robjects as robjects
from rpy2.robjects import r, pandas2ri, numpy2ri
from rpy2.robjects.conversion import localconverter

print("🚀 LANCEMENT DE LA PRODUCTION (100 MODÈLES) 🚀")

# ==============================================================================
# 1. CONFIGURATION ET DONNÉES
# ==============================================================================
# Config R
user_lib_path = os.path.abspath(os.path.expanduser("~/R_libs"))
r(f'.libPaths(c("{user_lib_path}", .libPaths()))')
r('library("gamlss")')

# Patch rpy2
def do_nothing(): pass
numpy2ri.activate = do_nothing
pandas2ri.activate = do_nothing
numpy2ri.deactivate = do_nothing
pandas2ri.deactivate = do_nothing

# Vérification Données
if 'df_pynm' not in locals():
    raise NameError("df_pynm manquant.")
if 'group' not in df_pynm.columns:
    df_pynm["group"] = df_pynm["diagnosis"].map({"CONTROL": 0, "SCHZ": 1})

# ==============================================================================
# 2. TRANSFERT DES DONNÉES (UNE SEULE FOIS)
# ==============================================================================
print("1. Transfert du dataset vers R...")
with localconverter(robjects.default_converter + pandas2ri.converter):
    r.assign("donnees_globales", df_pynm)

# On prépare le data.frame côté R une bonne fois pour toutes
r('df_global <- as.data.frame(donnees_globales)')
# On prépare le sous-ensemble d'entraînement côté R
r('df_train_global <- df_global[df_global$group == 0, ]')

print(f"   -> Dataset R prêt. Training set size: {r('nrow(df_train_global)')[0]}")

# ==============================================================================
# 3. BOUCLE D'ENTRAÎNEMENT
# ==============================================================================
print(f"2. Calcul des Z-scores pour {len(cols_pca)} composantes...")
warnings.filterwarnings('ignore')

df_deviations = pd.DataFrame(index=df_pynm.index)
df_deviations["diagnosis"] = df_pynm["diagnosis"]
if "site_id" in df_all.columns:
     df_deviations["site_id"] = df_all["site_id"].values[df_pynm.index]

success_count = 0

for pc_col in tqdm(cols_pca):
    try:
        # Script R optimisé : il utilise les objets globaux 'df_global' et 'df_train_global'
        # Plus besoin de transfert de données, c'est instantané.
        script_R = f"""
        # 1. Fit sur le subset pré-calculé
        # On utilise as.formula pour gérer dynamiquement le nom de la colonne
        fmla <- as.formula("{pc_col} ~ age + c(gender)")
        
        m <- gamlss(fmla, sigma.formula=~1, family=NO, data=df_train_global, trace=FALSE)
        
        # 2. Predict sur le dataset complet
        mu <- predict(m, newdata=df_global, what="mu", type="response")
        sigma <- predict(m, newdata=df_global, what="sigma", type="response")
        
        # 3. Export vecteurs
        res <- list(mu=as.vector(mu), sigma=as.vector(sigma))
        """
        
        # Exécution
        r(script_R)
        
        # Récupération résultats
        res_r = r['res']
        mu_vals = np.array(res_r.rx2('mu'))
        sigma_vals = np.array(res_r.rx2('sigma'))
        
        # Calcul Z-Score Python
        y_real = df_pynm[pc_col].values
        z_scores = (y_real - mu_vals) / sigma_vals
        
        df_deviations[f"z_{pc_col}"] = z_scores
        success_count += 1
        
    except Exception as e:
        # print(f"Erreur sur {pc_col}: {e}") # Décommenter si besoin
        df_deviations[f"z_{pc_col}"] = np.nan

# ==============================================================================
# 4. FIN
# ==============================================================================
print(f"\nTraitement terminé. Succès : {success_count}/{len(cols_pca)}")

output_file = "deviations_gamlss_pca100_final.csv"
df_deviations.to_csv(output_file, index=False)
print(f"✅ Résultats sauvegardés dans : {output_file}")

display(df_deviations.head())
```
